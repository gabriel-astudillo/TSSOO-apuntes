```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Configuración de la figura y ejes
fig, ax = plt.subplots(figsize=(10, 6.5))

tasks = ['Paralelo\n(2 Core)', 'Concurrente\n(1 Core)', 'Secuencial\n(1 Core)']

# --- Configuración de Tiempos (BLOQUES) ---

# Secuencial: T1 (0-4), luego T2 (4-8) - Ahora divididos verticalmente
seq_blocks = [
    {'start': 0, 'dur': 4, 'color': '#4C72B0', 'y': 3.12, 'label': 'T1'},
    {'start': 4, 'dur': 4, 'color': '#DD8452', 'y': 2.88, 'label': 'T2'}
]

# Concurrente: Intercalado alternado 
conc_t1 = [(0, 1), (2, 3), (4, 5), (6, 7)]  # T1
conc_t2 = [(1, 2), (3, 4), (5, 6), (7, 8)]  # T2
context_switches = [1, 2, 3, 4, 5, 6, 7] # Puntos de cambio

# Paralelo: Simultáneo (Posiciones 3.12 y 2.88)
par_blocks = [
    {'start': 0, 'dur': 4, 'color': '#4C72B0', 'y': 1.15, 'label': 'T1'},
    {'start': 0, 'dur': 4, 'color': '#DD8452', 'y': 0.85, 'label': 'T2'}
]

# --- Función de Actualización (FRAME A FRAME) ---

def update(frame):
    ax.clear()
    
    # Configuración base de los ejes
    ax.set_yticks([1, 2, 3])
    ax.set_yticklabels(tasks, fontsize=11, fontweight='bold')
    ax.set_xlim(0, 10)
    ax.set_ylim(0.4, 3.7)
    ax.set_xlabel('Tiempo', fontsize=11, fontweight='bold')
    ax.set_title('Comparativa de dinámica de ejecución\nSecuencial/Concurrente/Paralelo', fontsize=12, pad=15, fontweight='bold')
    ax.grid(axis='x', linestyle='--', alpha=0.5)

    t = frame / 10.0  # Tiempo simulado actual

    # === 1. SECUENCIAL (y ≈ 1) - AHORA DUAL ===
    for block in seq_blocks:
        if t > block['start']:
            current_dur = min(t - block['start'], block['dur'])
            # Dibujamos barras más delgadas (height=0.18) en dos alturas distintas
            ax.barh(block['y'], current_dur, left=block['start'], height=0.18, 
                    color=block['color'], edgecolor='black', alpha=0.85)

    # === 2. CONCURRENTE (y ≈ 2) - DUAL CON MARCAS ===
    # Dibujar marcas de Context Switch
    for cs in context_switches:
        if t >= cs:
            ax.vlines(x=cs, ymin=1.65, ymax=2.35, colors='#C44E52', linestyles=':', linewidth=1.2, alpha=0.7)
            ax.plot(cs, 2.0, marker='o', markersize=4, color='#C44E52')
            ax.text(1.0, 1.5, '┆ Cambio\nde Contexto', color='#C44E52', fontweight='bold', va='center', fontsize=8)

    # Sub-línea T1 (y=2.15)
    for start, end in conc_t1:
        dur = end - start
        if t > start:
            current_dur = min(t - start, dur)
            ax.barh(2.15, current_dur, left=start, height=0.18, color='#4C72B0', edgecolor='black', alpha=0.85)
            

    # Sub-línea T2 (y=1.85)
    for start, end in conc_t2:
        dur = end - start
        if t > start:
            current_dur = min(t - start, dur)
            ax.barh(1.85, current_dur, left=start, height=0.18, color='#DD8452', edgecolor='black', alpha=0.85)

    # === 3. PARALELO (y ≈ 3) - DUAL ===
    for block in par_blocks:
        if t > block['start']:
            current_dur = min(t - block['start'], block['dur'])
            ax.barh(block['y'], current_dur, left=block['start'], height=0.18, 
                    color=block['color'], edgecolor='black', alpha=0.85)

    # --- Leyendas explicativas ---
    ax.text(8.3, 2.0 + 1.15, '■ Thread 1', color='#4C72B0', fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 2.0 + 0.85, '■ Thread 2', color='#DD8452', fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 1.0 + 1.15, '■ Thread 1', color='#4C72B0', fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 1.0 + 0.85, '■ Thread 2', color='#DD8452', fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 1.15, '■ Thread 1', color='#4C72B0', fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 0.85, '■ Thread 2', color='#DD8452', fontweight='bold', va='center', fontsize=9)
    #ax.text(8.3, 2.0, '┆ Cambio\nde Contexto', color='#C44E52', fontweight='bold', va='center', fontsize=8)
    #ax.text(1.0, 1.5, '┆ Cambio\nde Contexto', color='#C44E52', fontweight='bold', va='center', fontsize=8)

# --- Ejecución y Guardado ---

# Crear la animación (85 cuadros, 10 frames por segundo simulados)
ani = animation.FuncAnimation(fig, update, frames=85, interval=100)

# Guardar como GIF animado compatible
ani.save('images/ejecucion_comparativa_4.gif', writer='pillow', fps=10)
plt.close()
```
