```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Crear figura y ejes
fig, ax = plt.subplots(figsize=(10, 6.5))

# Definición de tareas e intervalos de tiempo (start, end)
# Núcleo 1: T1 y T2 se intercalan
core1_t1 = [(0, 1), (2, 3), (4, 5), (6, 7)]
core1_t2 = [(1, 2), (3, 4), (5, 6), (7, 8)]

# Núcleo 2: T3 y T4 se intercalan
core2_t3 = [(0, 1), (2, 3), (4, 5), (6, 7)]
core2_t4 = [(1, 2), (3, 4), (5, 6), (7, 8)]

# Puntos de cambio de contexto
context_switches = [1, 2, 3, 4, 5, 6, 7]

# Paleta de colores diferenciada para las 4 tareas
colors = {
    'T1': '#3498db',  # Azul
    'T2': '#2ecc71',  # Verde
    'T3': '#e67e22',  # Naranja
    'T4': '#9b59b6',  # Púrpura
    'cs': '#e74c3c'   # Rojo para Context Switch
}

def update(frame):
    ax.clear()
    
    # Configuración del lienzo
    ax.set_yticks([1, 2])
    ax.set_yticklabels(['Core 1', 'Core 2'], fontsize=12, fontweight='bold')
    ax.set_xlim(0, 10)
    ax.set_ylim(0.4, 2.6)
    ax.set_xlabel('Tiempo', fontsize=11, fontweight='bold')
    ax.set_title('Ejemplo Paralelismo Concurrente\n4 Threads en 2 cores', fontsize=13, pad=15, fontweight='bold')
    ax.grid(axis='x', linestyle='--', alpha=0.5)

    t = frame / 10.0  # Tiempo simulado

    # --- Dibujar marcas de Context Switch ---
    for cs in context_switches:
        if t >= cs:
            # Marcador vertical para Núcleo 1
            ax.vlines(x=cs, ymin=0.75, ymax=1.25, colors=colors['cs'], linestyles=':', linewidth=1.2, alpha=0.7)
            ax.plot(cs, 1.0, marker='o', markersize=4, color=colors['cs'])
            
            # Marcador vertical para Núcleo 2
            ax.vlines(x=cs, ymin=1.75, ymax=2.25, colors=colors['cs'], linestyles=':', linewidth=1.2, alpha=0.7)
            ax.plot(cs, 2.0, marker='o', markersize=4, color=colors['cs'])

            ax.text(1.0, 1.5, '┆ Cambio\nde Contexto', color='#C44E52', fontweight='bold', va='center', fontsize=8)

    # === NÚCLEO 1 (y ≈ 1) ===
    # Tarea 1 (y = 1.12)
    for start, end in core1_t1:
        if t > start:
            dur = min(t - start, end - start)
            ax.barh(1.12, dur, left=start, height=0.18, color=colors['T1'], edgecolor='black', alpha=0.85)

    # Tarea 2 (y = 0.88)
    for start, end in core1_t2:
        if t > start:
            dur = min(t - start, end - start)
            ax.barh(0.88, dur, left=start, height=0.18, color=colors['T2'], edgecolor='black', alpha=0.85)

    # === NÚCLEO 2 (y ≈ 2) ===
    # Tarea 3 (y = 2.12)
    for start, end in core2_t3:
        if t > start:
            dur = min(t - start, end - start)
            ax.barh(2.12, dur, left=start, height=0.18, color=colors['T3'], edgecolor='black', alpha=0.85)

    # Tarea 4 (y = 1.88)
    for start, end in core2_t4:
        if t > start:
            dur = min(t - start, end - start)
            ax.barh(1.88, dur, left=start, height=0.18, color=colors['T4'], edgecolor='black', alpha=0.85)

    # --- Leyenda informativa lateral ---
    ax.text(8.3, 1.12, '■ Thread 1', color=colors['T1'], fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 0.88, '■ Thread 2', color=colors['T2'], fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 2.12, '■ Thread 3', color=colors['T3'], fontweight='bold', va='center', fontsize=9)
    ax.text(8.3, 1.88, '■ Thread 4', color=colors['T4'], fontweight='bold', va='center', fontsize=9)
    #ax.text(8.3, 1.50, '┆ Cambio\nde contexto', color=colors['cs'], fontweight='bold', va='center', fontsize=8)

# Generar animación
ani = animation.FuncAnimation(fig, update, frames=85, interval=100)

# Guardar GIF
ani.save('images/ejecucion_paralela_concurrente_1.gif', writer='pillow', fps=10)
plt.close()
```
