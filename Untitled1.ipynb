```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
```

```python
# Configuración de la figura y ejes
fig, ax = plt.subplots(figsize=(10, 6))

# Tareas y configuraciones
tasks = ['Secuencial', 'Concurrente (1 Núcleo)', 'Paralelo (2 Núcleos)']
ax.set_yticks([1, 2, 3])
ax.set_yticklabels(tasks, fontsize=12, fontweight='bold')
ax.set_xlim(0, 12)
ax.set_ylim(0.5, 3.5)
ax.set_xlabel('Tiempo (unidades)', fontsize=11)
ax.set_title('Diferencia entre Ejecución Secuencial, Concurrente y Paralela', fontsize=14, pad=15)
ax.grid(axis='x', linestyle='--', alpha=0.6)

# Definición de intervalos de ejecución (inicio, duración, color)
# Secuencial: Tarea 1 (0-4, azul), Tarea 2 (4-8, naranja)
seq_blocks = [(0, 4, 'skyblue', 'T1'), (4, 8, 'orange', 'T2')]

# Concurrente: Intercalado en 1 solo núcleo
conc_blocks = [
    (0, 1, 'skyblue', 'T1'), (1, 2, 'orange', 'T2'),
    (2, 3, 'skyblue', 'T1'), (3, 4, 'orange', 'T2'),
    (4, 5, 'skyblue', 'T1'), (5, 6, 'orange', 'T2'),
    (6, 7, 'skyblue', 'T1'), (7, 8, 'orange', 'T2')
]

# Paralelo: Dos núcleos simultáneos
par_blocks = [(0, 4, 'skyblue', 'T1 (Core 1)'), (0, 4, 'orange', 'T2 (Core 2)')]
```

```python
def update(frame):
    ax.clear()
    ax.set_yticks([1, 2, 3])
    ax.set_yticklabels(tasks, fontsize=12, fontweight='bold')
    ax.set_xlim(0, 10)
    ax.set_ylim(0.5, 3.5)
    ax.set_xlabel('Tiempo (unidades)', fontsize=11)
    ax.set_title('Diferencia entre Ejecución Secuencial, Concurrente y Paralela', fontsize=14, pad=15)
    ax.grid(axis='x', linestyle='--', alpha=0.6)
    
    t = frame / 10.0  # Tiempo actual de la simulación

    # 1. Dibujar Secuencial (y = 1)
    for start, dur, color, label in seq_blocks:
        if t > start:
            current_dur = min(t - start, dur)
            ax.barh(1, current_dur, left=start, height=0.4, color=color, edgecolor='black', alpha=0.85)

    # 2. Dibujar Concurrente (y = 2)
    for start, dur, color, label in conc_blocks:
        if t > start:
            current_dur = min(t - start, dur)
            ax.barh(2, current_dur, left=start, height=0.4, color=color, edgecolor='black', alpha=0.85)

    # 3. Dibujar Paralelo (y = 3)
    # Se desfasan un poco verticalmente para notar ambos núcleos
    for start, dur, color, label in par_blocks:
        if t > start:
            current_dur = min(t - start, dur)
            offset = 0.12 if 'Core 1' in label else -0.12
            ax.barh(3 + offset, current_dur, left=start, height=0.2, color=color, edgecolor='black', alpha=0.85)
```

```python
# Crear la animación (100 cuadros = 10 segundos)
ani = animation.FuncAnimation(fig, update, frames=85, interval=100)

# Guardar como GIF animado
ani.save('images/ejecucion_comparativa.gif', writer='imagemagick', fps=10)
plt.close()
```
